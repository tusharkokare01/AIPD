```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, classification_report
```

```python

df = pd.read_csv('IMDB Dataset.csv')
```

```python

df['sentiment'] = df['sentiment'].map({'positive': 1, 'negative': 0})


X_train, X_test, y_train, y_test = train_test_split(
    df['review'], df['sentiment'], test_size=0.2, random_state=42
)
```

```python

tfidf = TfidfVectorizer(stop_words='english', max_features=5000)
X_train_tfidf = tfidf.fit_transform(X_train)
X_test_tfidf = tfidf.transform(X_test)


model = LogisticRegression()
model.fit(X_train_tfidf, y_train)


y_pred = model.predict(X_test_tfidf)
```

```python
# Create a test sentence
test_review = ["This movie was a total worth of time, I love it."]

# Step 1: Convert text to numbers (Outcome starts here)
test_vector = tfidf.transform(test_review)

# Step 2: The Model makes a decision (The Prediction Outcome)
result = model.predict(test_vector)

# Step 3: Print the result
print("Positive" if result[0] == 1 else "Negative")
```

```python
print(f"Accuracy:  {accuracy_score(y_test, y_pred):.4f}")
print(f"Precision: {precision_score(y_test, y_pred):.4f}")
print(f"Recall:    {recall_score(y_test, y_pred):.4f}")
```
